```python

# ============================================================
# APEX | BALLAST STATION MANAGEMENT SYSTEM
# ADD / REMOVE / REPOSITION / MODIFY
# ============================================================

STATION_FIELDS = {
    "station_name",
    "x_m", "y_m", "z_m",
    "min_ballast_kg",
    "max_ballast_kg",
    "available",
    "safety_verified",
    "data_source"
}


def add_ballast_station(
    station_id,
    station_name,
    x_m,
    y_m,
    z_m,
    max_ballast_kg,
    min_ballast_kg=0.0,
    available=True
):
    global ballast_stations

    if station_id in ballast_stations["station_id"].values:
        raise ValueError("Station ID already exists.")

    new_station = {
        "station_id": station_id,
        "station_name": station_name,
        "x_m": x_m,
        "y_m": y_m,
        "z_m": z_m,
        "min_ballast_kg": min_ballast_kg,
        "max_ballast_kg": max_ballast_kg,
        "available": available,
        "safety_verified": False,
        "data_source": "User-defined assumption"
    }

    ballast_stations = pd.concat(
        [ballast_stations, pd.DataFrame([new_station])],
        ignore_index=True
    )

    validate_ballast_stations()

    print(f"Station {station_id} added successfully.")


def remove_ballast_station(station_id):
    global ballast_stations

    if station_id not in ballast_stations["station_id"].values:
        raise ValueError("Station ID not found.")

    ballast_stations = (
        ballast_stations[
            ballast_stations["station_id"] != station_id
        ].reset_index(drop=True)
    )

    print(f"Station {station_id} removed successfully.")


def update_ballast_station(station_id, **updates):
    global ballast_stations

    if station_id not in ballast_stations["station_id"].values:
        raise ValueError("Station ID not found.")

    invalid_fields = set(updates) - STATION_FIELDS

    if invalid_fields:
        raise ValueError(
            f"Invalid fields: {invalid_fields}"
        )

    idx = ballast_stations.index[
        ballast_stations["station_id"] == station_id
    ][0]

    for field, value in updates.items():
        ballast_stations.at[idx, field] = value

    validate_ballast_stations()

    print(f"Station {station_id} updated successfully.")


def validate_ballast_stations():
    df = ballast_stations

    required = [
        "station_id", "station_name",
        "x_m", "y_m", "z_m",
        "min_ballast_kg", "max_ballast_kg",
        "available", "safety_verified"
    ]

    assert not df[required].isnull().any().any(), (
        "Missing station data."
    )

    assert df["station_id"].is_unique, (
        "Duplicate station IDs."
    )

    numeric = [
        "x_m", "y_m", "z_m",
        "min_ballast_kg", "max_ballast_kg"
    ]

    values = df[numeric].to_numpy(dtype=float)

    assert np.isfinite(values).all(), (
        "Non-finite station values."
    )

    assert (df["z_m"] >= 0).all(), (
        "Ballast station cannot be below ground."
    )

    assert (df["min_ballast_kg"] >= 0).all(), (
        "Minimum ballast cannot be negative."
    )

    assert (
        df["max_ballast_kg"] >= df["min_ballast_kg"]
    ).all(), (
        "Maximum capacity must be >= minimum capacity."
    )

    assert (df["max_ballast_kg"] > 0).all(), (
        "Each station must have positive capacity."
    )

    assert df["available"].isin([True, False]).all()
    assert df["safety_verified"].isin([True, False]).all()

    return True


validate_ballast_stations()
print("APEX | BALLAST STATION VALIDATION PASSED")
```


# MODULE 02 | 3D VEHICLE PHYSICS ENGINE

This module calculates the combined mass properties of the
vehicle using component-level and ballast-level data.

The model includes:
- Total vehicle mass
- Three-dimensional centre of gravity
- Full inertia tensor
- Principal moments of inertia
- Static front and rear axle loads

Coordinate convention:
X = Longitudinal
Y = Lateral
Z = Vertical

All calculations use SI units.


```python

# ============================================================
# APEX | MODULE 02 - CELL 01
# 3D VEHICLE MASS PROPERTIES ENGINE
# ============================================================

def calculate_mass_properties(
    components,
    stations,
    ballast_allocation=None
):
    """
    Calculate combined vehicle mass properties.

    Parameters
    ----------
    components : DataFrame
        Vehicle component masses and XYZ coordinates.

    stations : DataFrame
        Ballast station locations and capacities.

    ballast_allocation : DataFrame or None
        Columns: station_id, ballast_kg

    Returns
    -------
    Dictionary containing total mass, 3D CG,
    inertia tensor and principal moments.
    """

    # Validate component data
    validate_ballast_stations()

    if components.empty:
        raise ValueError("Vehicle component database is empty.")

    if (components["mass_kg"] <= 0).any():
        raise ValueError("Component masses must be positive.")

    component_data = components[
        ["mass_kg", "x_m", "y_m", "z_m"]
    ].to_numpy(dtype=float)

    if not np.isfinite(component_data).all():
        raise ValueError("Invalid component numerical data.")

    masses = component_data[:, 0]
    positions = component_data[:, 1:4]

    # Add ballast if supplied
    if ballast_allocation is not None:

        required = {"station_id", "ballast_kg"}

        if not required.issubset(ballast_allocation.columns):
            raise ValueError("Invalid ballast allocation columns.")

        if ballast_allocation["station_id"].duplicated().any():
            raise ValueError("Duplicate station allocation IDs.")

        allocation = ballast_allocation.copy()

        if not np.isfinite(
            allocation["ballast_kg"].to_numpy(dtype=float)
        ).all():
            raise ValueError("Invalid ballast mass.")

        if (allocation["ballast_kg"] < 0).any():
            raise ValueError("Ballast cannot be negative.")

        station_lookup = stations.set_index("station_id")

        unknown = set(allocation["station_id"]) - set(station_lookup.index)

        if unknown:
            raise ValueError(f"Unknown stations: {unknown}")

        allocation = allocation.merge(
            stations[
                ["station_id", "x_m", "y_m", "z_m",
                 "min_ballast_kg", "max_ballast_kg", "available"]
            ],
            on="station_id",
            how="left",
            validate="one_to_one"
        )

        if (allocation["ballast_kg"] >
                allocation["max_ballast_kg"] + 1e-9).any():
            raise ValueError("Ballast exceeds station capacity.")

        if (allocation["ballast_kg"] <
                allocation["min_ballast_kg"] - 1e-9).any():
            raise ValueError("Ballast is below station minimum.")

        if (
            (~allocation["available"]) &
            (allocation["ballast_kg"] > 1e-9)
        ).any():
            raise ValueError("Ballast assigned to unavailable station.")

        active = allocation["ballast_kg"] > 0

        ballast_masses = allocation.loc[
            active, "ballast_kg"
        ].to_numpy(dtype=float)

        ballast_positions = allocation.loc[
            active, ["x_m", "y_m", "z_m"]
        ].to_numpy(dtype=float)

        masses = np.concatenate([masses, ballast_masses])
        positions = np.vstack([positions, ballast_positions])

    # Total mass
    total_mass = masses.sum()

    if total_mass <= 0:
        raise ValueError("Total vehicle mass must be positive.")

    # Combined 3D centre of gravity
    cg = np.sum(
        masses[:, None] * positions,
        axis=0
    ) / total_mass

    # Relative coordinates from combined CG
    relative_positions = positions - cg

    # Full inertia tensor using the parallel-axis contribution
    inertia_tensor = np.zeros((3, 3))

    for mass, (x, y, z) in zip(
        masses,
        relative_positions
    ):

        inertia_tensor += mass * np.array([
            [y**2 + z**2, -x*y, -x*z],
            [-x*y, x**2 + z**2, -y*z],
            [-x*z, -y*z, x**2 + y**2]
        ])

    # Principal moments and principal axes
    principal_moments, principal_axes = np.linalg.eigh(
        inertia_tensor
    )

    return {
        "total_mass_kg": total_mass,

        "cg_x_m": cg[0],
        "cg_y_m": cg[1],
        "cg_z_m": cg[2],

        "cg_coordinates_m": cg,

        "inertia_tensor_kg_m2": inertia_tensor,

        "Ixx_roll_kg_m2": inertia_tensor[0, 0],
        "Iyy_pitch_kg_m2": inertia_tensor[1, 1],
        "Izz_yaw_kg_m2": inertia_tensor[2, 2],

        "principal_moments_kg_m2": principal_moments,
        "principal_axes": principal_axes
    }


print("APEX | 3D PHYSICS ENGINE INITIALISED")
```

```python

# ============================================================
# APEX | MODULE 02 - CELL 03
# STATIC AXLE LOAD CALCULATOR
# ============================================================

def calculate_axle_loads(mass_kg, cg_x_m):

    wheelbase = VEHICLE_CONFIG["wheelbase_m"]
    gravity = VEHICLE_CONFIG["gravity_m_s2"]

    if not (0 <= cg_x_m <= wheelbase):
        raise ValueError(
            "CG lies outside the wheelbase. "
            "Check vehicle geometry and assumptions."
        )

    weight = mass_kg * gravity

    front_load_N = (
        weight * (wheelbase - cg_x_m) / wheelbase
    )

    rear_load_N = (
        weight * cg_x_m / wheelbase
    )

    front_percentage = front_load_N / weight * 100
    rear_percentage = rear_load_N / weight * 100

    return {
        "front_axle_load_N": front_load_N,
        "rear_axle_load_N": rear_load_N,
        "front_load_percentage": front_percentage,
        "rear_load_percentage": rear_percentage
    }


baseline_axle_results = calculate_axle_loads(
    baseline_results["total_mass_kg"],
    baseline_results["cg_x_m"]
)

print("APEX | BASELINE AXLE LOAD DISTRIBUTION")
print("-" * 45)

print(
    f"Front axle: "
    f"{baseline_axle_results['front_axle_load_N']:.2f} N"
)

print(
    f"Rear axle: "
    f"{baseline_axle_results['rear_axle_load_N']:.2f} N"
)

print(
    f"Front percentage: "
    f"{baseline_axle_results['front_load_percentage']:.2f}%"
)

print(
    f"Rear percentage: "
    f"{baseline_axle_results['rear_load_percentage']:.2f}%"
)
```

04 | BALLAST OPTIMISATION ENGINE.

```python

# ============================================================
# APEX | MODULE 03 - CELL 01
# OPTIMISATION CONFIGURATION
# ============================================================

from scipy.optimize import minimize, linprog

BASE_MASS = float(vehicle_components["mass_kg"].sum())

TARGET_MASS = VEHICLE_CONFIG["minimum_mass_kg"]

REQUIRED_BALLAST = max(0.0, TARGET_MASS - BASE_MASS)

TARGET_FRONT_FRACTION = 0.4525

FRONT_MIN = VEHICLE_CONFIG["front_load_min"]
FRONT_MAX = VEHICLE_CONFIG["front_load_max"]

print("=" * 55)
print("APEX | OPTIMISATION PARAMETERS")
print("=" * 55)

print(f"Baseline mass: {BASE_MASS:.2f} kg")
print(f"Target mass: {TARGET_MASS:.2f} kg")
print(f"Required ballast: {REQUIRED_BALLAST:.2f} kg")
print(f"Front load window: {FRONT_MIN*100:.1f}% - {FRONT_MAX*100:.1f}%")
print(f"Target front load: {TARGET_FRONT_FRACTION*100:.2f}%")

print("=" * 55)
```

```python

# ============================================================
# APEX | MODULE 03 - CELL 02
# FEASIBILITY CHECK
# ============================================================

available = ballast_stations["available"].to_numpy(dtype=bool)

active_stations = ballast_stations.loc[available].reset_index(drop=True)

station_x = active_stations["x_m"].to_numpy(dtype=float)

capacity = active_stations["max_ballast_kg"].to_numpy(dtype=float)

minimums = active_stations["min_ballast_kg"].to_numpy(dtype=float)

wheelbase = VEHICLE_CONFIG["wheelbase_m"]

base_x_moment = float(
    (
        vehicle_components["mass_kg"]
        * vehicle_components["x_m"]
    ).sum()
)

total_target_mass = BASE_MASS + REQUIRED_BALLAST

# Front load fraction:
# (L*M - total longitudinal mass moment) / (L*M)

upper_x_moment = (
    (1 - FRONT_MIN) * wheelbase * total_target_mass
    - base_x_moment
)

lower_x_moment = (
    (1 - FRONT_MAX) * wheelbase * total_target_mass
    - base_x_moment
)

# Linear programming feasibility constraints
A_ub = np.vstack([
    station_x,
    -station_x
])

b_ub = np.array([
    upper_x_moment,
    -lower_x_moment
])

bounds = list(zip(minimums, capacity))

feasibility_result = linprog(
    c=np.zeros(len(active_stations)),
    A_ub=A_ub,
    b_ub=b_ub,
    A_eq=np.ones((1, len(active_stations))),
    b_eq=[REQUIRED_BALLAST],
    bounds=bounds,
    method="highs"
)

print("=" * 55)
print("APEX | FEASIBILITY ANALYSIS")
print("=" * 55)

print("Solver status:", feasibility_result.message)

if feasibility_result.success:
    print("FEASIBLE CONFIGURATION EXISTS")
    print("Available capacity:", capacity.sum(), "kg")
    print("Required ballast:", REQUIRED_BALLAST, "kg")
else:
    raise RuntimeError(
        "No feasible ballast allocation exists. "
        "Review mounting capacities and vehicle constraints."
    )

print("=" * 55)
```

```python

# ============================================================
# APEX | MODULE 03 - CELL 03
# MULTI-OBJECTIVE FUNCTION
# ============================================================

OPTIMISATION_WEIGHTS = {
    "front_balance": 1.0,
    "lateral_balance": 0.2,
    "yaw_inertia": 0.05
}

baseline_inertia = baseline_results["Izz_yaw_kg_m2"]

if baseline_inertia <= 0:
    raise ValueError("Baseline yaw inertia must be positive.")


def ballast_objective(ballast_masses):

    allocation = pd.DataFrame({
        "station_id": active_stations["station_id"].to_numpy(),
        "ballast_kg": ballast_masses
    })

    properties = calculate_mass_properties(
        vehicle_components,
        active_stations,
        allocation
    )

    axle = calculate_axle_loads(
        properties["total_mass_kg"],
        properties["cg_x_m"]
    )

    front_error = (
        (axle["front_load_percentage"] / 100)
        - TARGET_FRONT_FRACTION
    ) / 0.01

    lateral_error = properties["cg_y_m"] / 0.05

    yaw_penalty = (
        properties["Izz_yaw_kg_m2"] / baseline_inertia
    )

    objective = (
        OPTIMISATION_WEIGHTS["front_balance"]
        * front_error**2

        + OPTIMISATION_WEIGHTS["lateral_balance"]
        * lateral_error**2

        + OPTIMISATION_WEIGHTS["yaw_inertia"]
        * yaw_penalty
    )

    return float(objective)


print("APEX | MULTI-OBJECTIVE FUNCTION READY")
print("Weights:", OPTIMISATION_WEIGHTS)
```

```python

# ============================================================
# APEX | MODULE 03 - CELL 04
# CONSTRAINED BALLAST OPTIMISATION
# ============================================================

def front_load_constraint(ballast_masses):

    allocation = pd.DataFrame({
        "station_id": active_stations["station_id"].to_numpy(),
        "ballast_kg": ballast_masses
    })

    properties = calculate_mass_properties(
        vehicle_components,
        active_stations,
        allocation
    )

    axle = calculate_axle_loads(
        properties["total_mass_kg"],
        properties["cg_x_m"]
    )

    return axle["front_load_percentage"] / 100


constraints = [

    # Exact ballast requirement
    {
        "type": "eq",
        "fun": lambda b: np.sum(b) - REQUIRED_BALLAST
    },

    # Lower front axle limit
    {
        "type": "ineq",
        "fun": lambda b: front_load_constraint(b) - FRONT_MIN
    },

    # Upper front axle limit
    {
        "type": "ineq",
        "fun": lambda b: FRONT_MAX - front_load_constraint(b)
    }
]

# Use a known feasible LP solution as the initial guess
initial_guess = feasibility_result.x.copy()

optimisation_result = minimize(
    fun=ballast_objective,
    x0=initial_guess,
    method="SLSQP",
    bounds=list(zip(minimums, capacity)),
    constraints=constraints,
    options={
        "maxiter": 2000,
        "ftol": 1e-10,
        "disp": True
    }
)

if not optimisation_result.success:
    raise RuntimeError(
        "Optimisation did not converge: "
        + optimisation_result.message
    )

optimal_ballast = optimisation_result.x

print("\nAPEX | OPTIMISATION COMPLETED")
print("Solver:", optimisation_result.message)
print("Iterations:", optimisation_result.nit)
print("Objective value:", optimisation_result.fun)
```

```python

# ============================================================
# APEX | MODULE 03 - CELL 05
# OPTIMAL BALLAST DISTRIBUTION
# ============================================================

optimal_allocation = pd.DataFrame({

    "station_id": active_stations["station_id"].to_numpy(),

    "station_name": active_stations["station_name"].to_numpy(),

    "ballast_kg": optimal_ballast

})

optimal_allocation["capacity_kg"] = capacity

optimal_allocation["utilisation_pct"] = (
    optimal_allocation["ballast_kg"]
    / optimal_allocation["capacity_kg"]
) * 100

print("=" * 65)
print("APEX | OPTIMAL BALLAST DISTRIBUTION")
print("=" * 65)

display(optimal_allocation.round(4))

print(
    "Total ballast:",
    f"{optimal_allocation['ballast_kg'].sum():.4f} kg"
)

print(
    "Remaining capacity:",
    f"{capacity.sum() - optimal_ballast.sum():.4f} kg"
)

print("=" * 65)
```

```python

# ============================================================
# APEX | MODULE 03 - CELL 06
# OPTIMISED VEHICLE VERIFICATION
# ============================================================

optimal_properties = calculate_mass_properties(
    vehicle_components,
    active_stations,
    optimal_allocation[["station_id", "ballast_kg"]]
)

optimal_axles = calculate_axle_loads(
    optimal_properties["total_mass_kg"],
    optimal_properties["cg_x_m"]
)

# Constraint checks
mass_pass = np.isclose(
    optimal_properties["total_mass_kg"],
    TARGET_MASS,
    atol=1e-6
)

front_percentage = optimal_axles["front_load_percentage"]

front_pass = (
    FRONT_MIN * 100 - 1e-6
    <= front_percentage
    <= FRONT_MAX * 100 + 1e-6
)

capacity_pass = np.all(
    optimal_ballast >= minimums - 1e-7
) and np.all(
    optimal_ballast <= capacity + 1e-7
)

ballast_pass = np.isclose(
    optimal_ballast.sum(),
    REQUIRED_BALLAST,
    atol=1e-6
)

# Recheck vertical force equilibrium
force_pass = np.isclose(
    optimal_axles["front_axle_load_N"]
    + optimal_axles["rear_axle_load_N"],
    TARGET_MASS * VEHICLE_CONFIG["gravity_m_s2"],
    atol=1e-6
)

print("=" * 60)
print("APEX | OPTIMISED CONFIGURATION VALIDATION")
print("=" * 60)

checks = {
    "Target vehicle mass": mass_pass,
    "Front axle load window": front_pass,
    "Ballast capacity limits": capacity_pass,
    "Total ballast requirement": ballast_pass,
    "Vertical force equilibrium": force_pass
}

for name, passed in checks.items():
    print(f"{name}: {'PASS' if passed else 'FAIL'}")

if not all(checks.values()):
    raise AssertionError("Optimised configuration failed validation.")

print("\nOPTIMISED VEHICLE PROPERTIES")
print(f"Total mass: {optimal_properties['total_mass_kg']:.3f} kg")
print(f"CG X: {optimal_properties['cg_x_m']:.4f} m")
print(f"CG Y: {optimal_properties['cg_y_m']:.4f} m")
print(f"CG Z: {optimal_properties['cg_z_m']:.4f} m")
print(f"Front axle: {front_percentage:.3f}%")
print(f"Rear axle: {optimal_axles['rear_load_percentage']:.3f}%")
print(f"Yaw inertia: {optimal_properties['Izz_yaw_kg_m2']:.3f} kg m²")

print("=" * 60)
```

```python

# ============================================================
# APEX | MODULE 03 - CELL 09
# MULTI-START OPTIMISATION ROBUSTNESS
# ============================================================

rng = np.random.default_rng(42)

multi_start_results = []

for run in range(25):

    # Generate a different feasible starting point
    random_cost = rng.uniform(0, 1, len(active_stations))

    lp_start = linprog(
        c=random_cost,
        A_ub=A_ub,
        b_ub=b_ub,
        A_eq=np.ones((1, len(active_stations))),
        b_eq=[REQUIRED_BALLAST],
        bounds=list(zip(minimums, capacity)),
        method="highs"
    )

    if not lp_start.success:
        continue

    result = minimize(
        fun=ballast_objective,
        x0=lp_start.x,
        method="SLSQP",
        bounds=list(zip(minimums, capacity)),
        constraints=constraints,
        options={
            "maxiter": 2000,
            "ftol": 1e-10
        }
    )

    if result.success:
        multi_start_results.append({
            "run": run + 1,
            "objective": result.fun,
            "iterations": result.nit,
            "allocation": result.x
        })

if not multi_start_results:
    raise RuntimeError("No multi-start run converged.")

multi_start_results.sort(
    key=lambda x: x["objective"]
)

best_run = multi_start_results[0]

print("=" * 60)
print("APEX | MULTI-START ROBUSTNESS REPORT")
print("=" * 60)

print("Successful runs:", len(multi_start_results))
print("Total attempted:", 25)

print(
    "Best objective:",
    f"{best_run['objective']:.10f}"
)

print(
    "Worst successful objective:",
    f"{multi_start_results[-1]['objective']:.10f}"
)

print(
    "Original objective:",
    f"{optimisation_result.fun:.10f}"
)

print(
    "Best run iterations:",
    best_run["iterations"]
)

print("=" * 60)
```

05 | DYNAMIC VEHICLE ANALYSIS

```python

# ============================================================
# APEX | MODULE 04 - CELL 01
# DYNAMIC SIMULATION CONFIGURATION
# ============================================================

DYNAMIC_CONFIG = {

    "front_roll_stiffness_fraction": 0.50,

    "front_track_m": VEHICLE_CONFIG["front_track_m"],

    "rear_track_m": VEHICLE_CONFIG["rear_track_m"],

    "gravity_m_s2": VEHICLE_CONFIG["gravity_m_s2"],

    "max_braking_deceleration_m_s2": 15.0,

    "max_lateral_acceleration_m_s2": 20.0
}

assert 0 <= DYNAMIC_CONFIG[
    "front_roll_stiffness_fraction"
] <= 1

print("APEX | DYNAMIC CONFIGURATION INITIALISED")
print("Front track:", DYNAMIC_CONFIG["front_track_m"], "m")
print("Rear track:", DYNAMIC_CONFIG["rear_track_m"], "m")
print(
    "Front roll stiffness share:",
    DYNAMIC_CONFIG["front_roll_stiffness_fraction"] * 100,
    "%"
)
```

```python

# ============================================================
# APEX | MODULE 04 - CELL 02
# LONGITUDINAL LOAD TRANSFER
# ============================================================

def calculate_longitudinal_load_transfer(
    mass_kg,
    cg_height_m,
    acceleration_m_s2,
    wheelbase_m
):

    gravity = DYNAMIC_CONFIG["gravity_m_s2"]

    if mass_kg <= 0 or wheelbase_m <= 0:
        raise ValueError("Mass and wheelbase must be positive.")

    if cg_height_m < 0:
        raise ValueError("CG height cannot be negative.")

    weight = mass_kg * gravity

    transfer = (
        mass_kg
        * acceleration_m_s2
        * cg_height_m
        / wheelbase_m
    )

    static_front = (
        optimal_axles["front_axle_load_N"]
    )

    static_rear = (
        optimal_axles["rear_axle_load_N"]
    )

    dynamic_front = static_front - transfer
    dynamic_rear = static_rear + transfer

    return {
        "static_front_N": static_front,
        "static_rear_N": static_rear,
        "dynamic_front_N": dynamic_front,
        "dynamic_rear_N": dynamic_rear,
        "load_transfer_N": transfer,
        "front_percentage": dynamic_front / weight * 100,
        "rear_percentage": dynamic_rear / weight * 100
    }


print("APEX | LONGITUDINAL LOAD TRANSFER ENGINE READY")
```

```python

# ============================================================
# APEX | MODULE 04 - CELL 03
# LATERAL LOAD TRANSFER
# ============================================================

def calculate_lateral_load_transfer(
    mass_kg,
    cg_height_m,
    lateral_acceleration_m_s2,
    front_axle_load_N,
    rear_axle_load_N
):

    front_share = DYNAMIC_CONFIG[
        "front_roll_stiffness_fraction"
    ]

    rear_share = 1 - front_share

    front_track = DYNAMIC_CONFIG["front_track_m"]
    rear_track = DYNAMIC_CONFIG["rear_track_m"]

    # Amount shifted from inner to outer side
    front_transfer = (
        front_share
        * mass_kg
        * abs(lateral_acceleration_m_s2)
        * cg_height_m
        / front_track
    )

    rear_transfer = (
        rear_share
        * mass_kg
        * abs(lateral_acceleration_m_s2)
        * cg_height_m
        / rear_track
    )

    # Positive acceleration is defined as load transfer
    # towards the positive-Y side of the vehicle.
    front_inner = front_axle_load_N / 2 - front_transfer
    front_outer = front_axle_load_N / 2 + front_transfer

    rear_inner = rear_axle_load_N / 2 - rear_transfer
    rear_outer = rear_axle_load_N / 2 + rear_transfer

    return {
        "front_inner_N": front_inner,
        "front_outer_N": front_outer,
        "rear_inner_N": rear_inner,
        "rear_outer_N": rear_outer,
        "front_transfer_N": front_transfer,
        "rear_transfer_N": rear_transfer
    }


print("APEX | LATERAL LOAD TRANSFER ENGINE READY")
```

```python

# ============================================================
# APEX | MODULE 04 - CELL 04
# COMBINED VEHICLE DYNAMICS
# ============================================================

def simulate_vehicle_dynamics(
    longitudinal_acceleration_m_s2,
    lateral_acceleration_m_s2
):

    mass = optimal_properties["total_mass_kg"]
    cg_height = optimal_properties["cg_z_m"]
    wheelbase = VEHICLE_CONFIG["wheelbase_m"]

    longitudinal = calculate_longitudinal_load_transfer(
        mass,
        cg_height,
        longitudinal_acceleration_m_s2,
        wheelbase
    )

    lateral = calculate_lateral_load_transfer(
        mass,
        cg_height,
        lateral_acceleration_m_s2,
        longitudinal["dynamic_front_N"],
        longitudinal["dynamic_rear_N"]
    )

    wheel_loads = {
        "Front inner": lateral["front_inner_N"],
        "Front outer": lateral["front_outer_N"],
        "Rear inner": lateral["rear_inner_N"],
        "Rear outer": lateral["rear_outer_N"]
    }

    total_wheel_load = sum(wheel_loads.values())

    expected_weight = (
        mass * DYNAMIC_CONFIG["gravity_m_s2"]
    )

    assert np.isclose(
        total_wheel_load,
        expected_weight,
        atol=1e-6
    ), "Dynamic vertical force equilibrium failed."

    return {
        "longitudinal": longitudinal,
        "lateral": lateral,
        "wheel_loads": wheel_loads,
        "total_vertical_load_N": total_wheel_load,
        "wheel_lift_detected": any(
            load <= 0 for load in wheel_loads.values()
        )
    }


print("APEX | COMBINED DYNAMICS ENGINE READY")
```

```python

# ============================================================
# APEX | MODULE 04 - CELL 05
# DYNAMIC SCENARIO TESTING
# ============================================================

scenarios = {

    "Straight line": {
        "longitudinal": 0.0,
        "lateral": 0.0
    },

    "Heavy braking": {
        "longitudinal": -8.0,
        "lateral": 0.0
    },

    "Cornering": {
        "longitudinal": 0.0,
        "lateral": 12.0
    },

    "Combined manoeuvre": {
        "longitudinal": -6.0,
        "lateral": 10.0
    }
}

scenario_results = {}

for name, conditions in scenarios.items():

    result = simulate_vehicle_dynamics(
        longitudinal_acceleration_m_s2=
            conditions["longitudinal"],

        lateral_acceleration_m_s2=
            conditions["lateral"]
    )

    scenario_results[name] = result

    print("\n" + "=" * 55)
    print("SCENARIO:", name.upper())
    print("=" * 55)

    print("Wheel loads (N):")

    for wheel, load in result["wheel_loads"].items():
        print(f"{wheel}: {load:.2f}")

    print(
        "Vertical equilibrium:",
        f"{result['total_vertical_load_N']:.2f} N"
    )

    print(
        "Wheel lift detected:",
        result["wheel_lift_detected"]
    )
```

```python

# ============================================================
# APEX | MODULE 04 - CELL 07
# BRAKING SENSITIVITY
# ============================================================

braking_levels = np.linspace(0, -15, 31)

braking_data = []

for acceleration in braking_levels:

    result = calculate_longitudinal_load_transfer(
        mass_kg=optimal_properties["total_mass_kg"],
        cg_height_m=optimal_properties["cg_z_m"],
        acceleration_m_s2=acceleration,
        wheelbase_m=VEHICLE_CONFIG["wheelbase_m"]
    )

    braking_data.append({
        "Braking deceleration (m/s²)": abs(acceleration),
        "Front axle (%)": result["front_percentage"],
        "Rear axle (%)": result["rear_percentage"]
    })

braking_df = pd.DataFrame(braking_data)

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=braking_df["Braking deceleration (m/s²)"],
    y=braking_df["Front axle (%)"],
    name="Front axle",
    mode="lines+markers",
    line=dict(color="#E31B23", width=3)
))

fig.add_trace(go.Scatter(
    x=braking_df["Braking deceleration (m/s²)"],
    y=braking_df["Rear axle (%)"],
    name="Rear axle",
    mode="lines+markers",
    line=dict(color="#F5C542", width=3)
))

fig.update_layout(
    title="APEX | BRAKING LOAD TRANSFER SENSITIVITY",
    template="plotly_dark",
    paper_bgcolor="#171719",
    plot_bgcolor="#242426",
    font_color="#F2EDE6",
    xaxis_title="Braking deceleration (m/s²)",
    yaxis_title="Axle load (%)",
    height=550
)

fig.show()
```

```python

# ============================================================
# APEX | MODULE 04 - CELL 08
# DYNAMIC PHYSICS VALIDATION
# ============================================================

mass = optimal_properties["total_mass_kg"]
weight = mass * DYNAMIC_CONFIG["gravity_m_s2"]

# 1. Vertical force equilibrium
for name, result in scenario_results.items():

    assert np.isclose(
        sum(result["wheel_loads"].values()),
        weight,
        atol=1e-6
    ), f"{name}: Force equilibrium failed."

# 2. No negative wheel loads in tested scenarios
for name, result in scenario_results.items():

    assert all(
        load >= 0
        for load in result["wheel_loads"].values()
    ), f"{name}: Negative wheel load detected."

# 3. Braking transfers load towards front
braking = scenario_results["Heavy braking"]["wheel_loads"]
static = scenario_results["Straight line"]["wheel_loads"]

assert (
    braking["Front inner"] + braking["Front outer"]
    >
    static["Front inner"] + static["Front outer"]
), "Braking front-load transfer failed."

# 4. Cornering transfers load towards outer wheels
cornering = scenario_results["Cornering"]["wheel_loads"]

assert (
    cornering["Front outer"] > cornering["Front inner"]
), "Front lateral transfer failed."

assert (
    cornering["Rear outer"] > cornering["Rear inner"]
), "Rear lateral transfer failed."

# 5. Straight-line lateral symmetry
assert np.isclose(
    static["Front inner"],
    static["Front outer"],
    atol=1e-6
)

assert np.isclose(
    static["Rear inner"],
    static["Rear outer"],
    atol=1e-6
)

print("=" * 60)
print("APEX | DYNAMIC PHYSICS VALIDATION")
print("=" * 60)

print("Vertical force equilibrium: PASS")
print("Non-negative wheel loads: PASS")
print("Braking load transfer: PASS")
print("Front lateral load transfer: PASS")
print("Rear lateral load transfer: PASS")
print("Straight-line symmetry: PASS")

print("=" * 60)
print("DYNAMIC MODEL VALIDATION PASSED")
```

```python

# ============================================================
# APEX | MODULE 06 - CELL 07
# LIVE DYNAMIC PHYSICS ENGINE
# ============================================================

def calculate_live_dynamics(
    properties,
    axle,
    longitudinal_acceleration,
    lateral_acceleration,
    front_roll_share
):

    mass = properties["total_mass_kg"]
    cg_height = properties["cg_z_m"]

    wheelbase = VEHICLE_CONFIG["wheelbase_m"]
    gravity = DYNAMIC_CONFIG["gravity_m_s2"]

    front_track = DYNAMIC_CONFIG["front_track_m"]
    rear_track = DYNAMIC_CONFIG["rear_track_m"]

    # Static axle loads from current editable configuration
    static_front = axle["front_axle_load_N"]
    static_rear = axle["rear_axle_load_N"]

    # Longitudinal transfer:
    # Negative acceleration = braking = load moves forwards
    longitudinal_transfer = (
        mass
        * longitudinal_acceleration
        * cg_height
        / wheelbase
    )

    dynamic_front = static_front - longitudinal_transfer
    dynamic_rear = static_rear + longitudinal_transfer

    # Lateral load transfer
    front_share = front_roll_share / 100.0
    rear_share = 1.0 - front_share

    front_transfer = (
        front_share
        * mass
        * abs(lateral_acceleration)
        * cg_height
        / front_track
    )

    rear_transfer = (
        rear_share
        * mass
        * abs(lateral_acceleration)
        * cg_height
        / rear_track
    )

    # Four wheel vertical loads
    wheel_loads = {
        "Front inner": dynamic_front / 2 - front_transfer,
        "Front outer": dynamic_front / 2 + front_transfer,
        "Rear inner": dynamic_rear / 2 - rear_transfer,
        "Rear outer": dynamic_rear / 2 + rear_transfer
    }

    expected_weight = mass * gravity
    calculated_weight = sum(wheel_loads.values())

    return {
        "mass_kg": mass,
        "longitudinal_transfer_N": longitudinal_transfer,
        "front_transfer_N": front_transfer,
        "rear_transfer_N": rear_transfer,
        "front_axle_N": dynamic_front,
        "rear_axle_N": dynamic_rear,
        "wheel_loads": wheel_loads,
        "expected_weight_N": expected_weight,
        "calculated_weight_N": calculated_weight,
        "equilibrium_error_N": calculated_weight - expected_weight,
        "wheel_lift": [
            name for name, load in wheel_loads.items()
            if load <= 0
        ]
    }


print("APEX | LIVE DYNAMIC PHYSICS ENGINE READY")
```

07.01 | CONSTRAINED BALLAST OPTIMISATION

```python

# ============================================================
# APEX | MODULE 07 - CELL 10
# CONSTRAINED BALLAST OPTIMISATION ENGINE
# ============================================================

from scipy.optimize import minimize

# Engineering targets
APEX_TARGET_MASS = float(TARGET_MASS)
APEX_FRONT_MIN = float(FRONT_MIN)
APEX_FRONT_MAX = float(FRONT_MAX)

def get_current_station_geometry():

    updated = apex_station_data.copy().reset_index(drop=True)

    for i, station in updated.iterrows():

        sid = station["station_id"]
        controls = ballast_controls[sid]

        updated.loc[i, "x_m"] = controls["x"].value
        updated.loc[i, "y_m"] = controls["y"].value
        updated.loc[i, "z_m"] = controls["z"].value

    return updated


def get_station_bounds(stations):

    bounds = []

    for _, station in stations.iterrows():

        lower = float(station.get("min_ballast_kg", 0.0))

        upper = float(
            station.get(
                "max_ballast_kg",
                station.get("capacity_kg", 20.0)
            )
        )

        if lower < 0 or upper < lower:
            raise ValueError(
                f"Invalid ballast capacity at {station['station_id']}"
            )

        bounds.append((lower, upper))

    return bounds


def evaluate_ballast_distribution(masses, stations):

    allocation = pd.DataFrame({
        "station_id": stations["station_id"].tolist(),
        "ballast_kg": np.asarray(masses, dtype=float)
    })

    properties = calculate_mass_properties(
        apex_vehicle_data,
        stations,
        allocation
    )

    axle = calculate_axle_loads(
        properties["total_mass_kg"],
        properties["cg_x_m"]
    )

    return allocation, properties, axle


def apex_optimisation_objective(masses, stations):

    _, props, _ = evaluate_ballast_distribution(
        masses, stations
    )

    # Normalised engineering objectives
    lateral_penalty = (props["cg_y_m"] / 0.10) ** 2

    height_penalty = (props["cg_z_m"] / 0.50) ** 2

    yaw_penalty = (
        props["Izz_yaw_kg_m2"] / 500.0
    )

    return (
        4.0 * lateral_penalty
        + 1.0 * height_penalty
        + 0.5 * yaw_penalty
    )


def optimise_apex_ballast():

    stations = get_current_station_geometry()
    bounds = get_station_bounds(stations)

    fixed_mass = float(apex_vehicle_data["mass_kg"].sum())

    required_ballast = APEX_TARGET_MASS - fixed_mass

    minimum_capacity = sum(b[0] for b in bounds)
    maximum_capacity = sum(b[1] for b in bounds)

    if not minimum_capacity <= required_ballast <= maximum_capacity:
        raise ValueError(
            f"Infeasible ballast requirement: "
            f"{required_ballast:.2f} kg required; "
            f"available range is "
            f"{minimum_capacity:.2f}–{maximum_capacity:.2f} kg."
        )

    # Start with the current ballast allocation
    initial = np.array([
        ballast_controls[sid]["mass"].value
        for sid in stations["station_id"]
    ], dtype=float)

    # Project starting allocation into feasible bounds
    initial = np.clip(
        initial,
        [b[0] for b in bounds],
        [b[1] for b in bounds]
    )

    # Ensure initial guess satisfies mass conservation
    remaining = required_ballast - initial.sum()

    for i, (lower, upper) in enumerate(bounds):

        if abs(remaining) < 1e-9:
            break

        available = (
            upper - initial[i]
            if remaining > 0
            else initial[i] - lower
        )

        adjustment = np.sign(remaining) * min(
            abs(remaining), available
        )

        initial[i] += adjustment
        remaining -= adjustment

    if abs(remaining) > 1e-6:
        raise ValueError("Could not construct a feasible initial allocation.")

    # Front axle constraints
    def front_lower_constraint(masses):

        _, _, axle = evaluate_ballast_distribution(
            masses, stations
        )

        return axle["front_load_percentage"] - APEX_FRONT_MIN * 100

    def front_upper_constraint(masses):

        _, _, axle = evaluate_ballast_distribution(
            masses, stations
        )

        return APEX_FRONT_MAX * 100 - axle["front_load_percentage"]

    constraints = [
        {
            "type": "eq",
            "fun": lambda masses: masses.sum() - required_ballast
        },
        {
            "type": "ineq",
            "fun": front_lower_constraint
        },
        {
            "type": "ineq",
            "fun": front_upper_constraint
        }
    ]

    result = minimize(
        fun=lambda masses: apex_optimisation_objective(
            masses, stations
        ),
        x0=initial,
        method="SLSQP",
        bounds=bounds,
        constraints=constraints,
        options={
            "maxiter": 1000,
            "ftol": 1e-10,
            "disp": False
        }
    )

    if not result.success:
        raise RuntimeError(
            "APEX optimisation failed: " + result.message
        )

    allocation, properties, axle = evaluate_ballast_distribution(
        result.x, stations
    )

    # Independent feasibility checks
    assert np.isclose(
        properties["total_mass_kg"],
        APEX_TARGET_MASS,
        atol=1e-5
    )

    front = axle["front_load_percentage"]

    assert (
        APEX_FRONT_MIN * 100 - 1e-4
        <= front
        <= APEX_FRONT_MAX * 100 + 1e-4
    )

    return {
        "allocation": allocation,
        "properties": properties,
        "axle": axle,
        "result": result,
        "stations": stations
    }


print("APEX | CONSTRAINED OPTIMISATION ENGINE READY")
```

```python

# ============================================================
# APEX | MODULE 09 - CELL 17
# MULTI-SCENARIO ENGINEERING ANALYSIS
# ============================================================

APEX_SCENARIOS = {
    "Straight line": (0.0, 0.0),
    "Acceleration": (4.0, 0.0),
    "Moderate braking": (-4.0, 0.0),
    "Heavy braking": (-8.0, 0.0),
    "Left cornering": (0.0, 12.0),
    "Right cornering": (0.0, -12.0),
    "Combined braking and cornering": (-6.0, 10.0)
}


def run_apex_scenario_analysis():

    stations, allocation, props, axle = (
        calculate_live_configuration()
    )

    records = []

    for scenario, (long_acc, lat_acc) in APEX_SCENARIOS.items():

        result = calculate_live_dynamics(
            properties=props,
            axle=axle,
            longitudinal_acceleration=long_acc,
            lateral_acceleration=lat_acc,
            front_roll_share=roll_share_control.value
        )

        row = {
            "Scenario": scenario,
            "Longitudinal (m/s²)": long_acc,
            "Lateral (m/s²)": lat_acc,
            "Front axle (N)": result["front_axle_N"],
            "Rear axle (N)": result["rear_axle_N"],
            "Front inner (N)": result["wheel_loads"]["Front inner"],
            "Front outer (N)": result["wheel_loads"]["Front outer"],
            "Rear inner (N)": result["wheel_loads"]["Rear inner"],
            "Rear outer (N)": result["wheel_loads"]["Rear outer"],
            "Minimum wheel load (N)": min(
                result["wheel_loads"].values()
            ),
            "Wheel lift warning": bool(result["wheel_lift"])
        }

        records.append(row)

    results = pd.DataFrame(records)

    return results


apex_scenario_results = run_apex_scenario_analysis()

display(
    apex_scenario_results.round(3)
)

print("APEX | SCENARIO ANALYSIS COMPLETED")
```

```python

# ============================================================
# APEX | MODULE 12 - CELL 24
# SYSTEM-WIDE ENGINEERING VALIDATION
# ============================================================

def run_apex_final_validation():

    stations, allocation, props, axle = (
        calculate_live_configuration()
    )

    checks = {}

    # 1. Mass conservation
    checks["Finite vehicle mass"] = np.isfinite(
        props["total_mass_kg"]
    )

    checks["Positive vehicle mass"] = (
        props["total_mass_kg"] > 0
    )

    # 2. Component data integrity
    checks["Unique component IDs"] = (
        apex_vehicle_data["component_id"].is_unique
    )

    checks["Positive component masses"] = (
        (apex_vehicle_data["mass_kg"] > 0).all()
    )

    # 3. Ballast integrity
    checks["Unique station IDs"] = (
        stations["station_id"].is_unique
    )

    checks["Non-negative ballast"] = (
        (allocation["ballast_kg"] >= 0).all()
    )

    # 4. Station capacities
    bounds = get_station_bounds(stations)

    checks["Ballast within capacities"] = all(
        lower - 1e-6 <= mass <= upper + 1e-6
        for mass, (lower, upper) in zip(
            allocation["ballast_kg"], bounds
        )
    )

    # 5. CG validity
    checks["Finite CG coordinates"] = all(
        np.isfinite(props[key])
        for key in ["cg_x_m", "cg_y_m", "cg_z_m"]
    )

    # 6. Dynamic force equilibrium
    dynamics = calculate_live_dynamics(
        props,
        axle,
        braking_control.value,
        cornering_control.value,
        roll_share_control.value
    )

    checks["Vertical force equilibrium"] = np.isclose(
        dynamics["calculated_weight_N"],
        dynamics["expected_weight_N"],
        atol=1e-5
    )

    # 7. Optimisation feasibility, if executed
    if apex_last_optimisation is not None:

        optimal = apex_last_optimisation

        checks["Optimisation solver success"] = (
            optimal["result"].success
        )

        checks["Optimised target mass"] = np.isclose(
            optimal["properties"]["total_mass_kg"],
            APEX_TARGET_MASS,
            atol=1e-5
        )

        front = optimal["axle"]["front_load_percentage"]

        checks["Optimised front axle window"] = (
            APEX_FRONT_MIN * 100 - 1e-4
            <= front
            <= APEX_FRONT_MAX * 100 + 1e-4
        )

    print("=" * 65)
    print("APEX | FINAL ENGINEERING VALIDATION")
    print("=" * 65)

    for name, passed in checks.items():

        print(
            f"{name:<40} "
            f"{'PASS' if passed else 'FAIL'}"
        )

    passed_count = sum(checks.values())
    total_count = len(checks)

    print("-" * 65)
    print(f"Tests passed: {passed_count}/{total_count}")

    if passed_count == total_count:
        print("APEX VALIDATION: ALL CHECKS PASSED")
    else:
        print("APEX VALIDATION: REVIEW REQUIRED")

    print("=" * 65)

    return checks


apex_validation_results = run_apex_final_validation()
```
